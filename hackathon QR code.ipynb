```python
import uuid
import qrcode
from PIL import Image, ImageTk
import tkinter as tk
from tkinter import ttk, messagebox

# Function to generate QR code for user
def generate_user_qr():
    username = "Alice"
    email = "alice@example.com"
    phone = "9876543210"

    unique_id = str(uuid.uuid4())
    user_data = f"""
Traveller ID: {unique_id}
Name: {username}
Email: {email}
Phone: {phone}
"""

    # Generate QR
    qr = qrcode.QRCode(
        version=1,
        error_correction=qrcode.constants.ERROR_CORRECT_H,
        box_size=10,
        border=4
    )
    qr.add_data(user_data)
    qr.make(fit=True)
    img = qr.make_image(fill_color="black", back_color="white")

    # Save QR temporarily
    filename = "user_qr.png"
    img.save(filename)

    # Display QR in Tkinter window
    qr_img = Image.open(filename)
    qr_img = qr_img.resize((250, 250))  # Resize for GUI
    qr_photo = ImageTk.PhotoImage(qr_img)
    qr_label.config(image=qr_photo)
    qr_label.image = qr_photo

    # Show success message
    messagebox.showinfo("Success", f"QR code generated!\nUnique ID: {unique_id}")

# Function to handle dropdown selection
def dropdown_action(event):
    choice = dropdown.get()
    if choice == "PROFILE":
        generate_user_qr()
   # elif choice == "B":
     #   messagebox.showinfo("Info", "Option B selected")
   # elif choice == "C":
       # messagebox.showinfo("Info", "Option C selected")

# Create main Tkinter window
root = tk.Tk()
root.title("Traveller QR Generator")
root.geometry("400x400")

# Dropdown Menu
options = ["PROFILE",]
dropdown = ttk.Combobox(root, values=options)
dropdown.set("Select an option")
dropdown.bind("<<ComboboxSelected>>", dropdown_action)
dropdown.pack(pady=20)

# Label to display QR code
qr_label = tk.Label(root)
qr_label.pack(pady=20)

root.mainloop()
```
